```python
import os
import pandas as pd

dir_path = r"C:\Users\carlo\Downloads\AI\AI\dataset\dataset"

all_data = []

for root, dirs, files in os.walk(dir_path):

    # Get current folder name
    folder_name = os.path.basename(root)

    for file in files:

        if file.endswith('.txt'):

            file_path = os.path.join(root, file)

            print(file_path)

            df = pd.read_csv(
                file_path,
                delimiter=',',
                header=None,
                names=['Full Energy', 'Variance']
            )

            # Add folder name as a column
            df['Protein_name'] = folder_name

            all_data.append(df)

# Combine all dataframes
final_df = pd.concat(all_data, ignore_index=True)

print(final_df.head())
```

```python
from pathlib import Path
import pandas as pd

base_dir = Path(r"C:\Users\carlo\Downloads\AI\AI\dataset\dataset")

for path in base_dir.rglob("*.rout"): #This is reading all the folders inside of the loop 
    print(f"Reading {path}")

    with open(path, "r", encoding="utf-8", errors="replace") as f:
        lines = f.readlines() # this is reading the lines of the file with rout 

    headers = lines[6].rstrip("\n").split("\t") # This code is to split the data into the correct columns, with teh headers

    rows = []
    for line in lines[7:]:
        line = line.rstrip("\n")
        if not line.strip():
            continue

        parts = line.split("\t")

        if len(parts) < len(headers):
            parts += [""] * (len(headers) - len(parts))
        elif len(parts) > len(headers):
            parts = parts[:len(headers)]

        rows.append(parts)

    df = pd.DataFrame(rows, columns=headers) # Makes a dataframe

    # Save next to the .rout file
    csv_path = path.with_suffix(".csv") # Saves it to the path name with the file
    df.to_csv(csv_path, index=False)

    print(f"Saved {csv_path}") #Prints the name to double check it works.
```

```python
from io import StringIO
root_dir = Path(r"C:\Users\carlo\Downloads\AI\AI\dataset\dataset")

header1 = "NodeId1\tInteraction\tNodeId2\tDistance\tAngle\tAtom1\tAtom2\tDonor\tPositive\tCation\tOrientation\tModel"
header2 = "NodeId\tChain\tPosition\tResidue\tType\tDssp\tDegree\tBfactor_CA\tx\ty\tz\tpdbFileName\tModel"

for rout_file in root_dir.rglob("*.rout"):

    protein_name = rout_file.parent.name
    folder = rout_file.parent  # save into same folder

    try:
        with open(rout_file, "r", encoding="utf-8", errors="ignore") as f:
            lines = f.readlines()

        idx1 = next(i for i, line in enumerate(lines) if line.strip() == header1)
        idx2 = next(i for i, line in enumerate(lines) if line.strip() == header2)

        table1_text = "".join(lines[idx1:idx2])
        table2_text = "".join(lines[idx2:])

        df1 = pd.read_csv(StringIO(table1_text), sep="\t")
        df2 = pd.read_csv(StringIO(table2_text), sep="\t")

        # save inside same folder as .rout file
        df1.to_csv(folder / f"table1_{protein_name}.csv", index=False)
        df2.to_csv(folder / f"table2_{protein_name}.csv", index=False)

        print(f"Saved CSVs in {folder}")

    except Exception as e:
        print(f"Failed on {rout_file}: {e}")
```
